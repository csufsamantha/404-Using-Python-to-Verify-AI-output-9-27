# Check Figures

## National picture (weeks 1-32, Total U.S.)

```python
# Filtering main/source data to show only 'Total U.S.' sales made during Weeks 1-32 of each year
data2026=maindata2026[(maindata2026['Geography']=='Total U.S.')&(maindata2026['Period']<=32)]
data2025=maindata2025[(maindata2025['Geography']=='Total U.S.')&(maindata2025['Period']<=32)]
```

```python
## CHECK ROW: Volume (units)
volume_change=data2026['Total Bulk and Bags'].sum()/data2025['Total Bulk and Bags'].sum()-1

print("Volume 2025:",f"{data2025['Total Bulk and Bags'].sum()/1e9:.3f}B")
print("Volume 2026:",f"{data2026['Total Bulk and Bags'].sum()/1e9:.3f}B")
print("Volume Change:",f"{volume_change:.1%}")
```

```python
## [X] CHECK ROW: Average price (ASP)
ASP_change=data2026['ASP Current Year'].sum()/data2025['ASP Current Year'].sum()-1

print("ASP 2025:",f"${data2025['ASP Current Year'].mean():.3f}")
print("ASP 2026:",f"${data2026['ASP Current Year'].mean():.3f}")
print("ASP Change:",f"{ASP_change:.1%}")
```

```python
# Revenue is estimated as weekly ASP x volume
data2026['Revenue']=data2026['ASP Current Year']*data2026['Total Bulk and Bags']
data2025['Revenue']=data2025['ASP Current Year']*data2025['Total Bulk and Bags']
```

```python
## CHECK ROW: Est. revenue
revenue_change=data2026['Revenue'].sum()/data2025['Revenue'].sum()-1

print("Revenue 2025:",f"${data2025['Revenue'].sum()/1e9:.3f}B")
print("Revenue 2026:",f"${data2026['Revenue'].sum()/1e9:.3f}B")
print("Revenue Change:",f"{revenue_change:.1%}")
```

```python
## CHECK ROW: Conventional revenue
convensional_2025=data2025[data2025['Type']!='Organic']['Revenue'].sum()
convensional_2026=data2026[data2026['Type']!='Organic']['Revenue'].sum()
convensional_change=convensional_2026/convensional_2025-1

print("Conventional revenue 2025:",f"${convensional_2025/1e9:.3f}B")
print("Conventional revenue 2026:",f"${convensional_2026/1e9:.3f}B")
print("Conventional revenue Change:",f"{convensional_change:.1%}")
```

```python
## CHECK ROW: Organic revenue
organic_2025=data2025[data2025['Type']=='Organic']['Revenue'].sum()
organic_2026=data2026[data2026['Type']=='Organic']['Revenue'].sum()
organic_change=organic_2026/organic_2025-1

print("Organic revenue 2025:",f"${organic_2025/1e6:.1f}M")
print("Organic revenue 2026:",f"${organic_2026/1e6:.1f}M")
print("Organic Revenue Change:",f"{organic_change:.1%}")
```

## Regional view

```python
# Filtering main/source data to show only sales made during Weeks 1-32 of each year
reg_data2026=maindata2026[(maindata2026['Period']<=32)]
reg_data2025=maindata2025[(maindata2025['Period']<=32)]
```

```python
## CHECK COLUMN: '2026 revenue'
reg_data2026['total_revenue2026']=reg_data2026['ASP Current Year']*reg_data2026['Total Bulk and Bags']
reg_revenue2026=reg_data2026.groupby('Geography')['total_revenue2026'].sum()

print("California:",f"${reg_revenue2026['California']/1e6:.1f}M")
print("West:",f"${reg_revenue2026['West']/1e6:.1f}M")
print("Northeast:",f"${reg_revenue2026['Northeast']/1e6:.1f}M")
print("Southeast:",f"${reg_revenue2026['Southeast']/1e6:.1f}M")
print("Midsouth:",f"${reg_revenue2026['Midsouth']/1e6:.1f}M")
print("Great Lakes:",f"${reg_revenue2026['Great Lakes']/1e6:.1f}M")
print("South Central:",f"${reg_revenue2026['South Central']/1e6:.1f}M")
print("Plains:",f"${reg_revenue2026['Plains']/1e6:.1f}M")
```

```python
## CHECK COLUMN: 'Volume'
reg_volume2026=reg_data2026.groupby('Geography')['Total Bulk and Bags'].sum()
reg_volume2025=reg_data2025.groupby('Geography')['Total Bulk and Bags'].sum()
reg_volume_change=reg_volume2026/reg_volume2025-1

print("California:",f"{reg_volume_change['California']:.1%}")
print("West:",f"{reg_volume_change['West']:.1%}")
print("Northeast:",f"{reg_volume_change['Northeast']:.1%}")
print("Southeast:",f"{reg_volume_change['Southeast']:.1%}")
print("Midsouth:",f"{reg_volume_change['Midsouth']:.1%}")
print("Great Lakes:",f"{reg_volume_change['Great Lakes']:.1%}")
print("South Central:",f"{reg_volume_change['South Central']:.1%}")
print("Plains:",f"{reg_volume_change['Plains']:.1%}")
```

```python
## [X] CHECK COLUMN: 'ASP'
reg_asp2026=reg_data2026.groupby('Geography')['ASP Current Year'].mean()
reg_asp2025=reg_data2025.groupby('Geography')['ASP Current Year'].mean()
reg_asp_change=reg_asp2026/reg_asp2025-1

print("California:",f"{reg_asp_change['California']:.1%}")
print("West:",f"{reg_asp_change['West']:.1%}")
print("Northeast:",f"{reg_asp_change['Northeast']:.1%}")
print("Southeast:",f"{reg_asp_change['Southeast']:.1%}")
print("Midsouth:",f"{reg_asp_change['Midsouth']:.1%}")
print("Great Lakes:",f"{reg_asp_change['Great Lakes']:.1%}")
print("South Central:",f"{reg_asp_change['South Central']:.1%}")
print("Plains:",f"{reg_asp_change['Plains']:.1%}")
```

```python
## CHECK COLUMN: 'Revenue"
reg_data2025['total_revenue2025']=reg_data2025['ASP Current Year']*reg_data2025['Total Bulk and Bags']
reg_revenue2025=reg_data2025.groupby('Geography')['total_revenue2025'].sum()
reg_revenue_change=reg_revenue2026/reg_revenue2025-1

print("California:",f"{reg_revenue_change['California']:.1%}")
print("West:",f"{reg_revenue_change['West']:.1%}")
print("Northeast:",f"{reg_revenue_change['Northeast']:.1%}")
print("Southeast:",f"{reg_revenue_change['Southeast']:.1%}")
print("Midsouth:",f"{reg_revenue_change['Midsouth']:.1%}")
print("Great Lakes:",f"{reg_revenue_change['Great Lakes']:.1%}")
print("South Central:",f"{reg_revenue_change['South Central']:.1%}")
print("Plains:",f"{reg_revenue_change['Plains']:.1%}")
```
